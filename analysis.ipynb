```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(x="Class", data=df)
plt.title("Fraud vs Normal Transactions")
plt.show()
```

```python
sns.countplot(x="Class", data=df)
plt.title("Fraud vs Normal Transactions")
plt.xlabel("Transaction Type (0 = Normal, 1 = Fraud)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Amount"], bins=50)
plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(x="Class", y="Amount", data=df)
plt.title("Fraud vs Normal Transaction Amounts")
plt.xlabel("Transaction Type (0 = Normal, 1 = Fraud)")
plt.ylabel("Amount")
plt.show()
```

```python
df["Time"].describe()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df["Time"], bins=50)
plt.title("Transaction Time Distribution")
plt.xlabel("Time (seconds)")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
sns.boxplot(x="Class", y="Time", data=df)
plt.title("Fraud vs Normal Transactions Over Time")
plt.xlabel("Transaction Type")
plt.ylabel("Time")
plt.show()
```

```python
fraud_counts = df['Class'].value_counts()

labels = ['Normal', 'Fraud']

plt.figure(figsize=(6,6))
plt.pie(fraud_counts, labels=labels, autopct='%1.2f%%')

plt.title("Fraud vs Normal Transactions")
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(data=df, x="Time", hue="Class", bins=50)

plt.title("Fraud vs Normal Transactions Over Time")

plt.show()
```
